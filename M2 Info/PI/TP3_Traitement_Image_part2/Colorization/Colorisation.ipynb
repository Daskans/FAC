## Transfering Color To Greyscale Images

Implémenter l'algorithme de l'article [Welsh et al., 2002]

Input : image à coloriser, image couleur de référence 
1. Convert source and target RGB images to lαβ color space.
2. Perform luminance remapping to match source and target histograms.
3. Precompute neighborhood statistics with a 5x5 pixel neighborhood.
4. For each greyscale pixel, find the best matching color sample and transfer α and β chromaticity values while retaining luminance.
